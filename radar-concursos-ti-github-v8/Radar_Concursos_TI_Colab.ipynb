# Radar de Concursos — painel único

Execute a célula abaixo uma vez. O fluxo principal é **âmbito → carreira → órgão → últimos concursos → cargo**.

Abas disponíveis:

1. Histórico por órgão;
2. Editais em aberto;
3. Pesquisa automática;
4. Adicionar PDF faltante;
5. Alertas;
6. Gerar/Publicar.

O PDF é apenas uma alternativa quando o concurso não foi localizado automaticamente.


```python
# Execute somente esta célula. Ela baixa/atualiza o projeto e abre o painel único.
import os, pathlib, shutil, subprocess, sys, importlib.util

missing = [name for name in ("ipywidgets", "pypdf") if importlib.util.find_spec(name) is None]
if missing:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *missing], check=True)

REPO_URL = "https://github.com/osmarrcs/radar-concursos-ti.git"
ROOT = pathlib.Path("/content/radar-concursos-ti")

if ROOT.exists() and (ROOT / ".git").exists():
    subprocess.run(["git", "-C", str(ROOT), "pull", "--ff-only"], check=True)
else:
    if ROOT.exists():
        shutil.rmtree(ROOT)
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(ROOT)], check=True)

os.chdir(ROOT)
sys.path.insert(0, str(ROOT / "src"))
from radar_concursos.colab_app import launch
APP = launch(ROOT)
```
